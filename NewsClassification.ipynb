```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_train = pd.read_csv('/kaggle/NewsClassification/Train.csv')
df_test = pd.read_csv('/kaggle/NewsClassification/Test.csv')
df_sub = pd.read_csv('/kaggle/NewsClassification/sample.csv')
```

```python
df_train
```

```python
df_test
```

```python
X = df_train['Headline']
y = df_train['Category']
X_test = df_test['Headline']
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.pipeline import Pipeline
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier()
```

```python
pipe = Pipeline([('tfidf', TfidfVectorizer()), ('model', model)])
```

```python
params = {
    'tfidf__max_features': [1000, 3000, 5000],
    'tfidf__ngram_range': [(1, 1), (1, 2), (1, 3)],
    'model__n_estimators': [100, 200],
    'model__max_depth': [10, 20, 30],
    'model__min_samples_split': [2, 5],
    'model__min_samples_leaf': [1, 2, 4]
}
```

```python
final_model = GridSearchCV(pipe, params, cv=3, verbose=2)
```

```python
final_model.fit(X, y)
```

```python
y_pred = final_model.predict(X_test)
```

```python
new_df = pd.DataFrame(columns = ['id', 'category'])
```

```python
new_df['id'] = np.arange(1,2001)
```

```python
new_df.drop('category', axis=1)
```

```python
new_df['category'] = y_pred
```
